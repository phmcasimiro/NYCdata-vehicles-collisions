## VARIÁVEIS (POO)
- No python `VARIÁVEIS` são consideradas `OBJETOS` e consequentemente as `VARIÁVEIS` têm `MÉTODOS` associados aos `TIPO DE VARIÁVEIS`
- **Exemplo:** Uma string "Pedro Henrique" pode ser manipulada de acordo com os métodos de string, por exemplo, `upper()`

## TIPOS E MÉTODOS DE VARIÁVEIS

- **IMPORTANTE:** No python não é necessário informar o tipo de variável. Se não declarar, ele declara automaticamente e você muda se quiser.
- **CUIDADO:** Preste atenção aos nomes restritos do python

| Tipo | Nome | Descrição | Exemplos de Métodos Comuns | O que o método faz? |
| :--- | :--- | :--- | :--- | :--- |
| `int` | Inteiro | Números sem decimais. | `bit_count()`, `as_integer_ratio()` | `bit_count()` retorna o número de bits 1 na representação binária. |
| `str` | String | Sequências de caracteres. | `upper()`, `split()`, `replace()`, `strip()` | `upper()` deixa tudo maiúsculo; `split()` divide o texto em uma lista. |
| `float` | Float | Números decimais. | `is_integer()`, `as_integer_ratio()` | `is_integer()` verifica se o decimal é um número inteiro (ex: 5.0). |
| `bool` | Booleano | Verdadeiro ou Falso. | `__int__()`, `__str__()` | Converte `True` para `1` ou `"True"` (possui poucos métodos próprios). |

### NOMES PERIGOSOS (Built-ins)

- Existem nomes que **não são proibidos**, mas você **não deve usar**, pois eles substituem funções nativas do Python e podem quebrar seu código:

*   **Tipos:** `int`, `str`, `float`, `list`, `dict`, `tuple`, `set`, `bool`
*   **Funções:** `print`, `input`, `id`, `type`, `len`, `sum`, `max`, `min`, `range`

*Exemplo do que NÃO fazer:*
```python
print = 10  # Agora você não consegue mais usar a função print() para exibir nada!
```

### TÉCNICAS DE PRINT()

````python
faturamento = 1000
custo = 600
lucro = faturamento - custo

# Estrutura de um print básico
print('#-------------------------------#')
print('#---------EXIBIR BÁSICO---------#')
print('#-------------------------------#')
print()
print('Relatório Financeiro da Empresa')
print('#-------------------------------#')

print('#---------------------------------------------------#')
print('#-----------------EXIBIR INTERMEDIÁRIO--------------#')
print('#---------------------------------------------------#')
print()
print('#---------MÉTODO DE CONCATENAÇÃO DE STRING COM VARIÁVEIS---------#')
# Por padrão o print() só exibe string, logo, precisamos transformar outras variáveis em string dentro do print
print('O custo da empresa foi de R$ ' + str(custo) + '.00 reais')
print()

print()
print('#---------------------------------#')
print('#---------EXIBIR AVANÇADO---------#')
print('#---------------------------------#')
print()
print('#----------MÉTODO FORMAT----------#')
print("O faturamento da empresa foi {} reais e o custo foi {} reais. Assim, o lucro da empresa foi {} reais.".format(faturamento, custo, lucro))
print()
print('#----------MÉTODO F-STRING----------#')
# A letra "f" antes do texto 'O faturamento...' indica um f-string, isto é, que a string é formatada
# Sem o "f" o python escreveria literamente "{faturamento:,.2f}"
# com o "f" o python entende que este trecho "{faturamento:,.2f}" é uma formatação
# As chaves "{}" delimitam o uso de uma variável no texto/string
# Os dois pontos ":" indicam que tudo que vem depois é uma regra de formatação
# A vírgula "," indica que o número deve usar a vírgula como separador de milhar
# O ponto "." indica que o separador decimal é o ponto
# o "2" indica a quantidade de casas decimais
# o "f" define a variável como float
print(f'O lucro foi R$ {faturamento-custo:,.2f}')

print('#---------------------------------#')
````

### MUDANÇA DE TIPO DE VARIÁVEL

````python
# Caixa de texto para o usuário digitar o faturamento
faturamento2 = float(input('Digite o faturamento da empresa: '))
# Caixa de texto para o usuário digitar o custo
custo2 = float(input('Digite o custo da empresa: '))
# Cálculo do lucro, transformando as variáveis faturamento2 e custo2 em float para realizar a subtração
lucro = faturamento2 - custo2
# Exibição do lucro usando f-string, onde o valor do lucro é formatado com vírgula como separador de milhar e 2 casas decimais.
print(f'O lucro da empresa foi de R$:{lucro:,.2f}') 
````
